```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
original = pd.read_csv("/content/original_dataset.csv", index_col=False)
```

```python
equivalent = original[original['RelationType']== 'Equivalent'][['RelationType','Governor','Dependent']]
equivalent = equivalent.rename({'RelationType':'relation', 'Governor': 'head', 'Dependent': 'tail'}, axis=1)
equivalent['relation'] = equivalent['relation'].replace({"Equivalent": "__label__Equivalent"})
```

```python
equivalent.head(5)
```

```python
equivalent.head()
```

```python
train, test = train_test_split(equivalent,
								   test_size=0.2,
								   random_state=42,
								   stratify=equivalent['relation'])
```

```python
test, val = train_test_split(test, test_size=0.5, random_state=42, stratify=test['relation'])
```

```python
train = train.dropna().drop_duplicates().reset_index(drop=True)
test = test.dropna().drop_duplicates().reset_index(drop=True)
val = val.dropna().drop_duplicates().reset_index(drop=True)
```

```python
test_df = pd.read_csv("/content/test_relations.tsv", index_col=False, sep="\t", names=['relation', 'head', 'tail'])
train_df = pd.read_csv("/content/train_relations.tsv", index_col=False, sep="\t", names=['relation', 'head', 'tail'])
dev_df = pd.read_csv("/content/dev_relations.tsv", index_col=False, sep="\t", names=['relation', 'head', 'tail'])

train_df[['head', 'relation', 'tail']].to_csv("/content/train_relations_new.tsv", sep="\t", index=False)
test_df[['head', 'relation', 'tail']].to_csv("/content/test_relations_new.tsv", sep="\t", index=False)
dev_df[['head', 'relation', 'tail']].to_csv("/content/dev_relations_new.tsv", sep="\t", index=False)
```

```python
test_df = pd.concat([test_df, test], axis=0).dropna().drop_duplicates().sample(frac=1).reset_index(drop=True)
train_df = pd.concat([train_df, train], axis=0).dropna().drop_duplicates().sample(frac=1).reset_index(drop=True)
dev_df = pd.concat([dev_df, val], axis=0).dropna().drop_duplicates().sample(frac=1).reset_index(drop=True)
```

```python
test_df = test_df[test_df['relation']!='__label__noRel']
train_df = train_df[train_df['relation']!='__label__noRel']
dev_df = dev_df[dev_df['relation']!='__label__noRel']
```

```python
test_df['relation'].value_counts()
```

```python
test_df.to_csv("/content/original_test_relations.tsv", sep="\t", index=False)
train_df.to_csv("/content/original_train_relations.tsv", sep="\t", index=False)
dev_df.to_csv("/content/original_dev_relations.tsv", sep="\t", index=False)
```

```python
dataset = pd.concat([train_df, test_df], axis=0)
```

```python
dataset = pd.concat([dataset, dev_df], axis=0)
```

```python
dataset = dataset.dropna().drop_duplicates().sample(frac=1).reset_index(drop=True)
```
